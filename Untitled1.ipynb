На первом этапе я загружаю разархивированные файлы с масками, изображениями и геоданными в класс, определяя путьк этим файлам.

На третьем этапе я определяю маски для изображений.

```python
class DSTLDataset(Dataset):
    def __init__(
        self,
        root_dir: str,
        csv_file: str,
        grid_sizes_file: str,
        img_size=(512, 512),
        reduce_factor: int = 1,
    ):
        super().__init__()
        self.root_dir = root_dir
        self.df = pd.read_csv(csv_file)
        self.grid_sizes = pd.read_csv(grid_sizes_file, names=["ImageId", "Xmax", "Ymin"], skiprows=1)
        self.image_ids = self.df['ImageId'].unique()[::reduce_factor]
        self.img_size = img_size
        self.classes = list(range(1, 11))  # 10 классов

        self.transform = A.Compose([
            A.Resize(img_size[0], img_size[1], interpolation=cv2.INTER_LINEAR),
            A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
            ToTensorV2(),
        ])
        self.mask_transform = A.Compose([
            A.Resize(img_size[0], img_size[1], interpolation=cv2.INTER_NEAREST),
            ToTensorV2(),
        ])

    #Возвращает размер датасета
    def __len__(self):
        return len(self.image_ids)

    #Функция для получения элемента массива
    def __getitem__(self, idx):
        image_id = self.image_ids[idx]
        img_path = os.path.join(self.root_dir, f"{image_id}.tif")

        # Загружаем изображение
        with rasterio.open(img_path) as src:
            img = src.read([1, 2, 3]).transpose(1, 2, 0).astype(np.uint8)
            h, w = img.shape[:2]

        # Получаем масштабные коэффициенты
        xmax = float(self.grid_sizes[self.grid_sizes.ImageId == image_id].Xmax.iloc[0])
        ymin = float(self.grid_sizes[self.grid_sizes.ImageId == image_id].Ymin.iloc[0])
        x_scale = w / xmax
        y_scale = h / ymin

        # Функция масштабирования WKT → пиксели
        def scale_poly(poly):
            return transform(lambda x, y: (x * x_scale, y * y_scale), poly)

        # Создание маски
        mask = np.zeros((h, w), dtype=np.uint8)
        sub = self.df[self.df['ImageId'] == image_id]

        for class_id in self.classes:
            polys = sub[sub['ClassType'] == class_id]['MultipolygonWKT']
            for wkt_str in polys:
                if wkt_str != 'MULTIPOLYGON EMPTY':
                    poly = wkt.loads(wkt_str)
                    poly = scale_poly(poly)
                    mask += rasterio.features.rasterize(
                        [(poly, class_id)],
                        out_shape=(h, w),
                        fill=0,
                        all_touched=True,
                        dtype=np.uint8
                    )

        # Аугментация изображения
        transformed = self.transform(image=img)
        img_tensor = transformed["image"]

        # Маска (приводим к размеру и в тензор)
        mask_resized = cv2.resize(mask, self.img_size, interpolation=cv2.INTER_NEAREST)
        mask_tensor = torch.from_numpy(mask_resized).long()

        return img_tensor, mask_tensor
```

В данной работе используется архитектура U-Net, хорошо зарекомендовавшая себя в задаче сегментации. Усли выв хотиет узнать больше подробностей, рекомендую прочитать статью на хабре:https://habr.com/ru/articles/746842/

```python
class DoubleConv(nn.Module):
  def __init__(self, in_channels, out_channels):
    super().__init__()
    self.double_conv = nn.Sequential(
        nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1),
        nn.BatchNorm2d(out_channels),
        nn.ReLU(inplace=True),
        nn.Conv2d(out_channels, out_channels, kernel_size=3, padding=1),
        nn.BatchNorm2d(out_channels),
        nn.ReLU(inplace=True)
    )

  def forward(self, x):
    return self.double_conv(x)

class UNet(nn.Module):
  def __init__(self, n_classes = 10):
      super().__init__()

      # Encoder
      self.inc = DoubleConv(3, 64)
      self.down1 = Down(64, 128)
      self.down2 = Down(128, 256)
      self.down3 = Down(256, 512)
      self.down4 = Down(512, 1024)
      self.down5 = Down(1024, 2048)
      # Decoder
      self.up0 = Up(2048, 1024)
      self.up1 = Up(1024, 512)
      self.up2 = Up(512, 256)
      self.up3 = Up(256, 128)
      self.up4 = Up(128, 64)

      # Output
      self.outc = nn.Conv2d(64, n_classes, kernel_size =1)

      # Построение архитектуры нейронной сети
  def forward(self,x):
    x1 = self.inc(x)
    x2 = self.down1(x1)
    x3 = self.down2(x2)
    x4 = self.down3(x3)
    x5 = self.down4(x4)
    x6 = self.down5(x5)

    x = self.up0(x6, x5)
    x = self.up1(x, x4)
    x = self.up2(x, x3)
    x = self.up3(x, x2)
    x = self.up4(x, x1)

    return self.outc(x)

class Down(nn.Module):
  def __init__(self, in_channels, out_channels) -> None:
     super().__init__()
     self.maxpool_conv = nn.Sequential(
         nn.MaxPool2d(2),
         nn.ReLU(inplace=True),
         DoubleConv(in_channels, out_channels),
         nn.BatchNorm2d(out_channels)
     )

  def forward(self, x):
    return self.maxpool_conv(x)

class Up(nn.Module):
  def __init__(self, in_channels, out_channels) -> None:
     super().__init__()
     self.up = nn.ConvTranspose2d(in_channels, in_channels // 2, kernel_size=2, stride=2)
     nn.BatchNorm2d(out_channels)
     self.conv = DoubleConv(in_channels, out_channels)

  def forward(self, x1, x2):
    x1 = self.up(x1)
    diffY = x2.size()[2] - x1.size()[2]
    diffX = x2.size()[3] - x1.size()[3]

    x1 = F.pad(x1, [diffX // 2, diffX - diffX // 2,
                    diffY // 2, diffY - diffY // 2])
    x = torch.cat([x2, x1], dim =1)
    return self.conv(x)
```

Пишем процесс обучения нейронной сети

```python
def dice_coeff(pred, target, smooth=1e-8):
    # Вычисляем Dice для одного класса
    intersection = (pred * target).sum()
    union = pred.sum() + target.sum()
    return (2. * intersection + smooth) / (union + smooth)

def multiclass_dice_coeff(preds, targets, num_classes=11, ignore_background=True):
    """
    Вычисляет средний Dice Coefficient по всем классам.

    Args:
        preds: (N, H, W) - предсказанные классы (argmax по каналу)
        targets: (N, H, W) - истинные маски
        num_classes: общее количество классов
        ignore_background: игнорировать ли фон (класс 0)
    """
    total_dice = 0.0
    count = 0
    start_class = 1 if ignore_background else 0

    for class_id in range(start_class, num_classes):
        # Создаем бинарные маски для текущего класса
        pred_binary = (preds == class_id).float()
        target_binary = (targets == class_id).float()

        # Пропускаем классы, отсутствующие в ground truth
        if target_binary.sum() == 0:
            continue

        class_dice = dice_coeff(pred_binary, target_binary)
        total_dice += class_dice
        count += 1

    return total_dice / count if count > 0 else 1.0

def train_model(model, train_loader, val_loader, optimizer, scheduler, device, epochs):
    for epoch in range(epochs):
        model.train()
        train_loss = 0.0
        train_dice = 0.0

        for inputs, masks in train_loader:
            inputs = inputs.to(device)
            masks = masks.to(device)

            optimizer.zero_grad()
            outputs = model(inputs)

            # Вычисляем loss (кросс-энтропия)
            loss = F.cross_entropy(outputs, masks)
            loss.backward()
            optimizer.step()

            # Вычисляем предсказанные классы
            preds = outputs.argmax(dim=1)

            # Вычисляем Dice
            dice = multiclass_dice_coeff(preds, masks)

            train_loss += loss.item()
            train_dice += dice.item()

        # Валидация
        model.eval()
        val_loss = 0.0
        val_dice = 0.0

        with torch.no_grad():
            for inputs, masks in val_loader:
                inputs = inputs.to(device)
                masks = masks.to(device)

                outputs = model(inputs)
                loss = F.cross_entropy(outputs, masks)
                preds = outputs.argmax(dim=1)
                dice = multiclass_dice_coeff(preds, masks)

                val_loss += loss.item()
                val_dice += dice.item()

        # Логирование
        print(f"Epoch {epoch+1}/{epochs}")
        print(f"Train Loss: {train_loss/len(train_loader):.4f} | Train Dice: {train_dice/len(train_loader):.4f}")
        print(f"Val Loss: {val_loss/len(val_loader):.4f} | Val Dice: {val_dice/len(val_loader):.4f}")

        if scheduler:
            scheduler.step(val_loss)
```

Оптимизатор

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu') # Повторно проверяем есть ли у нас GPU
model = UNet(n_classes=19).to(device)
optimizer = optim.AdamW(model.parameters(), lr=1e-4, weight_decay=1e-5) # Взят AdamW, хорошо показыл себя
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'max', patience=3, factor=0.5)

train_model(model, train_loader=train_dataloader, val_loader=test_dataloader, optimizer=optimizer, scheduler = scheduler, device=device, epochs=10) # 10 эпох достаточно
```

Левое значение - кросс-эентропия. Правое - оценка для масок.

```python
def denormalize(tensor, mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]):
    """Денормализует тензор изображения для визуализации"""
    tensor = tensor.clone()
    for t, m, s in zip(tensor, mean, std):
        t.mul_(s).add_(m)  # t = t * s + m
    return tensor.clamp(0, 1)


def visualize_prediction(model, dataloader, device, idx=1):
    model.eval()
    with torch.no_grad():
        images, masks = next(iter(dataloader))
        images = images.to(device)
        outputs = model(images)
        preds = torch.argmax(F.softmax(outputs, dim=1), dim=1).cpu()

        img = denormalize(images[idx].cpu()).permute(1, 2, 0).numpy()
        true_mask = masks[idx].cpu().numpy()
        pred_mask = preds[idx].numpy()

        fig, axs = plt.subplots(1, 3, figsize=(18, 6))
        axs[0].imshow(img)
        axs[0].set_title("Входное изображение")
        axs[1].imshow(true_mask, cmap='jet')
        axs[1].set_title("Истинная маска")
        axs[2].imshow(pred_mask, cmap='jet')
        axs[2].set_title("Предсказанная маска")
        plt.show()
```

```python
visualize_prediction(model, test_dataloader, device, idx=0)
```
